```python
df['salary']=df['salary'].str.lower().str.strip()
```

```python
df['salary'] = df['salary'].replace(['N/A', 'NA', 'n/a', 'na'], np.nan)
```

```python
word_map = {
    'one lakh': 100000, 
    'fifty thousand': 50000
}

# Map text values to numbers (lowercasing strings for safe matching)
df['salary'] = df['salary'].replace(word_map)
```

```python
df['salary'] = (
    df['salary']
    .str.replace('Rs.', '', regex=False)
    .str.replace('Rs', '', regex=False)
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.rstrip('-') # Removes trailing dashes like '167043-'
    .str.strip()     # Removes leftover spaces
)
```

```python
# Step 1: Attempt to convert Join_Date to datetime, turning invalid text into NaT
parsed_dates = pd.to_datetime(df['join_date'], errors='coerce')
```

```python
# Step 2: Create a mask where the original value is not null, but the parsed date is null (meaning it's text/invalid)
text_entries_mask = df['join_date'].notna() & parsed_dates.isna()
```

```python
# Step 3: Extract and view the unique text or invalid entries found in the column
unique_text_values = df.loc[text_entries_mask, 'join_date'].unique()
```

```python
print("Non-date text entries found:")
print(unique_text_values)
```

```python
# Clean the column: converts valid dates to standard format, turns text/invalid into NaT
df['join_date'] = pd.to_datetime(df['join_date'], errors='coerce', format='mixed')

# Verify the changes
print(df['join_date'].head())
print(f"Total missing dates (NaT): {df['join_date'].isna().sum()}")
```

```python
# Apply backward fill to the Join_Date column
df['join_date'] = df['join_date'].bfill()

# Verify that all missing values have been filled
print("Null values after backward fill:", df['join_date'].isna().sum())
```

```python
# Convert the column to strings so we can manipulate the text safely
df['phone_number'] = df['phone_number'].astype(str)

# Step 1: Replace any text entries (like "Not Given" or "nan") with actual missing values
df.loc[df['phone_number'].str.contains(r'[a-zA-Z]', na=False), 'phone_number'] = np.nan

# Step 2: Remove anything that isn't a digit (like +, -, spaces)
df['phone_number'] = df['phone_number'].str.replace(r'\D', '', regex=True)

# Step 3: Replace numbers starting with '92' or '00' so they start with a standard '0'
df['phone_number'] = df['phone_number'].str.replace(r'^92', '0', regex=True)
df['phone_number'] = df['phone_number'].str.replace(r'^00', '0', regex=True)

# Step 4: Check length and keep only the ones that are exactly 11 digits long
valid_length_mask = df['phone_number'].str.len() == 11
df['phone_number'] = df['phone_number'].where(valid_length_mask, np.nan)

# Verify the clean results and check for missing values
print(df['phone_number'].head(10))
print(f"\nTotal missing or invalid numbers: {df['phone_number'].isna().sum()}")
```

```python
# Spaces remove karein
df['performance_score'] = df['performance_score'].astype('string').str.strip()

# Check karein kaun-si values directly numeric nahi hain
numeric_check = pd.to_numeric(
    df['performance_score'],
    errors='coerce'
)

text_values = df.loc[
    numeric_check.isna() & df['performance_score'].notna(),
    'performance_score'
].unique()

print("Categorical/Text values:")
print(text_values)
```

```python
# the above can also be done as


# Create a mask for alphabetical characters, ignoring 'NaN'
text_mask = df['performance_score'].astype(str).str.contains(r'[a-zA-Z]', na=False)
text_mask = text_mask & (df['performance_score'].astype(str).str.lower() != 'nan')

# Print the extracted text values
print(df.loc[text_mask, 'performance_score'].value_counts())
```

```python
# Define what numeric score each text value should represent
score_mapping = {
    'Excellent': 95,
    'Good': 75,
    'Average': 50,
    'Poor': 25
}
```

```python
# Replace the text strings in the column with their new numeric values
df['performance_score'] = df['performance_score'].replace(score_mapping)
```

```python
# Check the results to confirm the text has been successfully swapped for numbers
text_mask = df['performance_score'].astype(str).str.contains(r'[a-zA-Z]', na=False)
text_mask = text_mask & (df['performance_score'].astype(str).str.lower() != 'nan')

print("Text values remaining after mapping:")
print(df.loc[text_mask, 'performance_score'].value_counts())
```
